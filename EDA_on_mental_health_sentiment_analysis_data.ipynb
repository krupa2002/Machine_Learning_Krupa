## Visualize the frequency of different sentiments

Importing necesaary visualization library

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
from wordcloud import WordCloud # to generate a word cloud visualization.

text = " ".join(df['status'].astype(str)) # " " is to consider all the text , .join is to ensure all text data is joined, Ensures all values are strings using .astype(str) to avoid missing values

wordcloud = WordCloud(width=800, height=400, background_color='white').generate(text) #deciding size, background, 

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(8,5)) #to adjust figure size
sns.countplot(y=df["status"], order=df["status"].value_counts().index) #Counter plot, y axis as sentiments with descending order 
plt.title("Sentiment Distribution")
plt.show()
```
